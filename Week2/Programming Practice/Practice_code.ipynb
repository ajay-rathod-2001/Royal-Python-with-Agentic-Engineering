## 💥 Check if a Number is Positive and Negative:

### 👨‍🏫 Method-1: Using Brute Force

```python
num = int(input("Enter the any number : "))

if num > 0:
	print("Number is Positive")
elif num < 0:
	print("Number is Negative")
else:
	print('zero')
```

### 👨‍🏫 Method-2: Using Nested if-else Statements

```python
num =str(input("Enter the any Number : "))

if num >= str (0):
	if num == str(0):
		print('zero')
	else:
		print('Number is Positive')
else:
	print('Number is Negative')
```

### 👨‍🏫 Method-3: Using Ternary Operator

```python
num =int(input("Enter an any Number : "))

print("Positive" if num >= 0 else "Negative")
```

## 💥 Check the Whether a Number  is Even or Odd:

### 👨‍🏫 Method-1: Using Brute Force

```python
num = int(input("Enter a number : "))

if num % 2 == 0:
	print(f"This is the Even Number : {num}")
else:
	print(f"This is the Odd Number : {num}")
```

### 👨‍🏫 Method-2: Using Ternary operator

```python
num = int(input("Enter a Number: "))

print("Even:",num) if num % 2 ==0 else print("Odd:",num)

# print("Even:") if num % 2 ==0 else print("Odd:")
```

```python
num = int(input("Enter a Number: "))
print("Even") if num % 2 ==0 else print("Odd")
```

### 👨‍🏫 Method-3: Using Bitwise operator

```python
def isEven(num):
	return not num & 1

if __name__== "__main__":
	num = int(input("Enter Any Number: "))

	if isEven(num):
		print("Even:",num )
	else:
		print("Odd:",num)
```
