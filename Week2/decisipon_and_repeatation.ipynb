## 🌈💥 Nested Loops & Match / case (Mini Project):

### 👨‍🏫 Concept -1 : Nested Loop ( Loop per mai Loop)

```python
for i in range (5):
	for j in range(4):
		print("i =", i,"j =",j)
```

* i --> Define the --> Column
* j --> Define the --> Row

```python
for i in range(1,4):
	for j in range (1,4):
		print(f"{i}-{j}",end=" ")
	print()
```

```python
for i in range (4,7):
	for j in range (1,4):
		print(f"{i}*{j}", end=" ")
	print()
```

```python
for i in range (1,5):
	print("*"*i)
```

### 👨‍🏫 Concept-2 : match/Case (Saaf Rounting)

```python
while True:
	user_input = input("Enter a Command for software (start/stop/pause): ")
	match user_input:
		case "start":
			print("Software is Running.....!💥💫")
		case "stop":
			print("Software is Stop.....❌")
		case "pause":
			print("Software is Paused....⚡")
		case _:
			print("Invalid Command, Please Enter the Valid Command")

	sw_stop = input(f"Do you want to stop the programm? (yes/no)")

	if sw_stop == "yes":
		break


```
