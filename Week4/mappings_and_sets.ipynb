## 🌈💞 Comprehensions :

### 👨‍🏫 Concept-1: Old Methods Vs New Methods

* Old Methos:

```python
import time

start_time= time.time()

square = []
for num in range(1,7):
	square.append(num**num)
	# print(f"Power of {num} : {num**num}")
square
```

* New Method:

```python
sqr_with_com=[num**num for num in range(1,7)]
sqr_with_com
```

```python
start_time=time.time()

sqr_with_com=[num**num for num in range(1,7)]

end_time=time.time()
print(f"Total Time taken exicuted above the code: {end_time - start_time}")
```

## 🌈💝 Sorting:

### 👨‍🏫 Concept-1: Custom sorting sorted( key=....)

* Sort key:  A sort key is the feature passed as the key argument to sorted() or  .sort(). It is called on each  elment to produced a value is used  for comprenhension , alowing items to be derived property rather than their raw value.

```python
sorted([2,5,3,4,8,9,1,6])
```

```python
stundents = [
				{"name":"Ajay", "marks":89},
				{"name":"Jay", "marks":80},
				{"name":"Vijay", "marks":90},
				{"name":"sanjay", "marks":99}
]
```

```python
sorted(stundents, key =lambda s: s["marks"])
```

```python
stundents.sort( key =lambda s: s["marks"])
stundents
```
