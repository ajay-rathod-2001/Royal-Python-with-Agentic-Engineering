# 🌈🚩 The Flag / Search Pattern

### 👨‍🏫 Concept-1: "Kya list mein 8 hai?"

```python
numbs=[2,4,6,9,1,8,7,10]
find_value = 8
found = False

for num in numbs:
	if num == find_value:
		found= True
print(f"{find_value} is present : {found}")
```

## 🔀🏷️ Mini Project: Create FizzBuzz Game

```python
end_user_value = int(input("Enter the end value of FizzBuzz Game:"))

for i in range(1,end_user_value):
	if i%3 == 0 and i%5 == 0:
		print("FizzBuzz")
	elif i%3 == 0:
		print("Fizz")
	elif i%5 == 0:
		print("Buzz")
	else:
		print(i)
```
