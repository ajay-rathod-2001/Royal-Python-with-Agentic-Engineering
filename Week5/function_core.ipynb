## 💞💥 *args / **kwargs

### 👨‍🏫 Concept-1: *args (jitne marzi values)

```python
def sum_all(*a):
	return sum(a)
```

```python
sum_all(1,4,6,2)
```

```python
def my_bio(*args):
	print(f"Name:{args[0]}, Age: {args[1]}, City: {args[2]}")
```

```python
my_bio("Ai",32, "Beed")
```

```python
my_bio("Ai", "Beed", 32) ## argument position is matter
```

### 👨‍🏫 Concept-2: **kwargs (jitne marzi naam-vali values)

```python
def my_bio(**kwargs):
	for key, value in kwargs.items():
		print(f"{key}: {value}")
```

```python
my_bio(
			name="john", 
			age=34,
			city="mumbai",
			salary=40000 )
```

### 👨‍🏫 Concept-3: sab ek saath (order yaad rakho)

```python
def stud_info(paper_marks,*marks, **info):
	total_marks= sum(marks)
	percentage=total_marks/(len(marks)*paper_marks) * 100

	for key, value in info.items():
		print(f"{key}: {value}")

	print("Total Marks: ",total_marks)
	print("Percentage of: ",percentage)


# obtain_marks=[55,77,88,93,90]
# stud_information={
# 						"name":"Shyam",
# 						"age":29,
# 						"city":"Pune"
# }

stud_info(
						100,55,77,88,93,90,
						name="Shyam",
						age=29,
						city="Pune"
		)
```

## 🌈💥 The Mutable : Default Trap & Sentinels

```python
def add_item(item,cart=[]):
	cart.append(item)
	return cart
```

```python
add_item("Mango")
```

```python
add_item("Apple")
```

```python
def add_item(item,cart=None):
	if cart is None:
		cart=[]
		cart.append(item)
		return cart
```

```python
add_item("Santra")
```

```python
add_item("Mosambi")
```

### 👨‍🏫 Concept-2: "diya hi nahi" vs "None diya" (Sentinel idea)

```python
object()
```

```python
__MISSING = object()

def update(value=__MISSING ):
	if value is __MISSING :
		print("No value was given at all")
	elif value is None:
		print("User passed the None value directly.")
	else:
		print(f"User gave : {value}")
```

```python
update()
```

```python
update("apple")
```

### 🏷️ Demo: Safe history appender

```python
def llm_prompts(text,history=None):
	if history is None:
		history= []
	history.append({"role":"user", "context": text})
	return history
```

```python
llm_prompts("What is llm?")
```
