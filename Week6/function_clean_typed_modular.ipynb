## 🌈💞 Recursion :

```python
def countdown(n):
	print(n)

	if n == 1:
		print("countdown completed")
		return
		
	countdown(n-1)
```

```python
countdown(9)
```

### 👨‍🏫 Concept-1 : Factorial

```python
def factorial(n):

	if n ==1:
		return 1

	return n * factorial(n-1)
```

```python
factorial(5)
```

### 👨‍🏫 Conecpt-3 : Nested data

```python
def sum_next_list(list_data):
	total = 0

	for item in list_data:
		
		if isinstance(item, list):
			
			total = total + sum_next_list(item)
		else:
			total = total + item

	return total 
```

```python
sum_next_list([2, 5, 8,[1, 5, 10, [3,6,7]], 12, 5, 14])
```

### 👨‍🏫 Concept-3 : Dono ko saath maein (Professional fuction)

```python
def cal_area(length: float , width: float) -> float:
	"""
		Calculation the area of Rectangle:

		Args:
				length : The lenght of the rectangle
				width : The width of the rectangle
		Returns:
				The area of reactangle

	"""

	return float(length * width )
```
