### 👨‍🏫 Concept-3: Validation (Methods galat data rokte hain )

```python
class BankAccount:
	"""
		A Class to represent a bank account:

		Attributes:
				balance (int): The balance of the bank account.
				
		Methods:
				Deposite(Amount: int) -> None: Deposite money into the account.
				withdraw(Amount: int) -> NOne: Withdraw money from the account.
				get_balance() -> int: Get the current Balance of the account.


	"""


	def __init__(self, balance: int):
		self._balance = balance

	def deposite(self, amount: int ) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")  

		if amount > 0:
			self._balance += amount 
		return print("Available Balance is ", self._balance)

	def withdraw(self, amount: int) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")    

		if amount > 0:
			self._balance -= amount
		return print("Available Balance is ", self._balance)

	def get_balance(self) -> int:
		return self._balance
```

```python
bank1 = BankAccount(10000)
```

```python
bank1.deposite(5000)
```

```python
bank1.withdraw(7000)
```

```python
bank1.deposite(-5000)
```

```python
bank1._balance = 400000
```

```python
bank1._balance
```

### 👨‍🏫 Concept-4: __ (double underscore) - aur strict --> private plus security provides

```python
class BankAccount:
	"""
		A Class to represent a bank account:

		Attributes:
				balance (int): The balance of the bank account.
				
		Methods:
				Deposite(Amount: int) -> None: Deposite money into the account.
				withdraw(Amount: int) -> NOne: Withdraw money from the account.
				get_balance() -> int: Get the current Balance of the account.


	"""


	def __init__(self, balance: int):
		self.__balance = balance

	def deposite(self, amount: int ) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")  

		if amount > 0:
			self.__balance += amount 
		return print("Available Balance is ", self.__balance)

	def withdraw(self, amount: int) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")  

		if amount > self.__balance:
			return print("Insufficient Balance..!")

		if amount > 0:
			self.__balance -= amount
		return print("Available Balance is ", self.__balance)

	def get_balance(self) -> int:
		return self.__balance 
```

```python
bank_Acc = BankAccount(12000)
```

```python
bank_Acc.__balance
```

```python
bank_Acc.get_balance()
```

```python
bank_Acc.withdraw(1000)
```

```python
bank_Acc.get_balance()
```

```python
bank_Acc.withdraw(12000)
```

```python
bank_Acc.withdraw(-12000)
```

# 🌈💥 Inheritance :

### 👨‍🏫 Concept-1: Basic inheritance

```python
class Animal:
	def __init__(self, name):
		self.name = name

	def eat(self):
		print(f"{self.name} is eating")

	def sleep(self):
		print(f"{self.name} is sleeping")
```

```python
cow_animal = Animal("Cow")
cow_animal.sleep()
```

```python
class Dog(Animal):
	def bark(self):
		print(f"{self.name} is barking")
```

```python
dog = Dog("Tommy")
```

```python
dog.bark()
```

```python
dog.eat()
```

```python
class A:
	pass
class B(A):
	pass
class C(B):
	pass
class D(C):
	pass

class F(A,B):
	pass
class G(A, B, C):
	pass
class H(A, B, C, D):
	pass
```

### 👨‍🏫 Concept-3: Super() ( parent ka setup use karo))

```python
class Dog(Animal):
	def __init__(self, name, breed_name):
		super().__init__(name)
		self.breed_name = breed_name

	def bark(self):
		print(f"{self.name} is Barking")

	def breed(self):
		print(f"{self.name} is a {self.breed_name}")
```

```python
dog_9 = Dog("Tommy", "Labrador")
```

```python
dog_9.breed()
```

```python
dog_9.breed_name
```

```python
dog_9.name
```

```python
dog_9.bark()
```

# 🌈🔀 Polymorphism :

- Poly = Many, morph = shapes

### 👨‍🏫 Concept-1 : Method Overrinding (child parent ko badle)

```python
class Animal:
	def speak(self):
		print("Some generic sound comming")

class Dog(Animal):
	def speak(self):
		print("Bhoouuu Bhoouuu...!")

class Cat(Animal):
	def speak(self):
		print("Meow Meow...!")

class Cow(Animal):
	def speak(self):
		print("Ambyaaa Ambyaaa...!")
```

```python
dogg = Dog()
dogg.speak()
```

```python
Animal().speak()
```

```python
Cow().speak()
```

```python
Cat().speak()
```

### 👨‍🏫 Concept-2 : Asli Power: same code , alag Objects

```python
animals = [Dog(), Cat(), Cow()]

for animal in animals:
	animal.speak()
```

### 👨‍🏫 Concept - 2 : Abstact Base Class (ABC)

```python
from abc import ABC, abstractmethod 
```

```python
class Tool(ABC):  # Abstract class

	@abstractmethod
	def area(self):
		...

```

```python
class Rectangle(Tool):
	def __init__(self, length, width):
		self.length = length
		self.width = width

	def area(self):
		return self.length * self.width

class Circle(Tool):
	def area(self, radius):
		return 3.14 * radius * radius

class Triangle(Tool):
	def area(self, base, height):
		return 0.5 * base * height

class Cylinder(Tool):
	def area(self, radius, height):
		return 2 * 3.14 * radius * height   
```

```python
Rectangle(6, 5).area()
```

```python
Circle().area(10)
```

```python
Triangle().area(7, 12)
```

```python
Cylinder().area(5, 7)
```
