### 👨‍🏫 Concept-3: Validation (Methods galat data rokte hain )

```python
class BankAccount:
	"""
		A Class to represent a bank account:

		Attributes:
				balance (int): The balance of the bank account.
				
		Methods:
				Deposite(Amount: int) -> None: Deposite money into the account.
				withdraw(Amount: int) -> NOne: Withdraw money from the account.
				get_balance() -> int: Get the current Balance of the account.


	"""


	def __init__(self, balance: int):
		self._balance = balance

	def deposite(self, amount: int ) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")  

		if amount > 0:
			self._balance += amount 
		return print("Available Balance is ", self._balance)

	def withdraw(self, amount: int) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")    

		if amount > 0:
			self._balance -= amount
		return print("Available Balance is ", self._balance)

	def get_balance(self) -> int:
		return self._balance
```

```python
bank1 = BankAccount(10000)
```

```python
bank1.deposite(5000)
```

```python
bank1.withdraw(7000)
```

```python
bank1.deposite(-5000)
```

```python
bank1._balance = 400000
```

```python
bank1._balance
```

### 👨‍🏫 Concept-4: __ (double underscore) - aur strict --> private plus security provides

```python
class BankAccount:
	"""
		A Class to represent a bank account:

		Attributes:
				balance (int): The balance of the bank account.
				
		Methods:
				Deposite(Amount: int) -> None: Deposite money into the account.
				withdraw(Amount: int) -> NOne: Withdraw money from the account.
				get_balance() -> int: Get the current Balance of the account.


	"""


	def __init__(self, balance: int):
		self.__balance = balance

	def deposite(self, amount: int ) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")  

		if amount > 0:
			self.__balance += amount 
		return print("Available Balance is ", self.__balance)

	def withdraw(self, amount: int) -> None:
		if amount <= 0:
			return print("Invalid Amount..!")  

		if amount > self.__balance:
			return print("Insufficient Balance..!")

		if amount > 0:
			self.__balance -= amount
		return print("Available Balance is ", self.__balance)

	def get_balance(self) -> int:
		return self.__balance 
```

```python
bank_Acc = BankAccount(12000)
```

```python
bank_Acc.__balance
```

```python
bank_Acc.get_balance()
```

```python
bank_Acc.withdraw(1000)
```

```python
bank_Acc.get_balance()
```

```python
bank_Acc.withdraw(12000)
```

```python
bank_Acc.withdraw(-12000)
```
