### 👨‍🏫 Concept-2: −−str−− (insaano ke liye sundar)

* Definition - −−str−− and −−repr−− :- These are the special ("dunder") method that define an object's text form.
* −−str−− :- Returns a readable string for end users ( used by print() and str() ), while 
* −−repr−− :- Returns an unambiguous, developer-orinted string for debugging ( used by repr() and in containers). If only one is defined, −−repr−− acts as the fallback

```python
class Dog:
	def __init__(self, name, age):
		self.name = name
		self.age = age

	def __str__(self):
		return f"Dog: {self.name}, {self.age} Year old"
d = Dog("Buddy", 3)
print(d)
```

### 👨‍🏫 Concept-3: −−repr−− (developer ke liye)

```python
class Dog:
	def __init__(self, name, age):
		self.name = name
		self.age = age

	def __str__(self):
		return f"Dog: {self.name}, {self.age} Year old"

	def __repr__(self) :
				return f"Dog:(name = {self.name}, age = {self.age} Year old )"

		
d = Dog("Buddy", 3)
print(d)
```

### 👨‍🏫 Concept-2 : ––eq–– ( apni equality define karo )

* Definition :- ––eq–– and ––len–– :- Dunder methods let object work built-in operations. 
* ––eq–– : Define how the == operator compare two objects (usually by comparing their data). 
* ––len–– : Define what the built-in len() returns for an object.

```python

class Point:
	def __init__(self, x, y) :
		self.x = x
		self.y = y 

	def __eq__(self, other):
		return self.x == other.x and self.y == other.y

p1 = Point(1, 2)
p2 = Point(1, 2)
p1 == p2
```

### 👨‍🏫 Concetp-2: From_dict (super common in AI)

```python
class Person:
	def __init__(self, name, age ) :
		self.name =name
		self.age = age

	@classmethod
	def from_dict(cls, data):
		name, age = data["name"], data["age"]
		return cls(name, age)
```

```python
class Person:
	def __init__(self, name, age ) :
		self.name =name
		self.age = age

	@classmethod
	def from_dict(cls, data):
		name, age = data["name"], data["age"]
		return cls(name, age)

	@classmethod
	def from_string(cls, data):
		name, age = data.split(",")
		return cls(name, int(age))

	def __repr__(self ) :
		return f"Person(name = {self.name}, age = {self.age})"

	def __str__(self ) :
		return f"Person: {self.name}, {self.age} year old"
```

```python
p1 = Person("Abhay", 21)
print(p1)
```

```python
p2 = Person.from_string("Abhay, 21")
print(p2)
```

```python
p3 = Person.from_dict({"name":"Abhayjeet", "age": 21})
print(p3)
```

# 🌈🏷️ Property :

### 👨‍🏫 Concept-1: Computed property (apane aap calculate)

* Definination - @property :- The @property decorator turns a method into a read-only attibute , that is accessed without parentheses and computed each time it is  read. Adding a matching  @<.name.>.setterthe value to be assigned while running logic, combining a simple attribute with controlled access. 

```python
class Circle:
	def __init__(self, radius) :
		self.radius = radius

	@property
	def area(self):
		return 3.14159 * self.radius ** 2
```

```python
circle = Circle(4)
circle.area
```

```python
circle.radius = 10 
circle.area
```

```python
class Temperature:
	def __init__(self, celsius = 0):
		self._celsius = celsius

	@property
	def celsius(self):
		return self._celsius

	@celsius.setter
	def celsius(self, value):
		if value < -273.15:
			raise ValueError("Below absolute zero!")
		
		self._celsius = value
```

```python
temp = Temperature()
```

```python
temp.celsius
```

```python
temp.celsius = 6
```

```python
temp.celsius
```

```python
temp.celsius = -275
```

# 🌈🧠 Composition :

* "Has-a" ---> Relationship .... rather than inheriting from them 
* "is-a" ---> Relationship .... collaborating parts and is more flexible than inheritance.

```python
class Engine:
	def __init__(self, horsepower):
		self.horsepower = horsepower

	def start(self):
		return f"{self.horsepower} HP Engine Started"

class Car:
	def __init__(self, brand, horsepower):
		self.brand = brand
		self.horsepower = horsepower
		self.engine = Engine(self.horsepower).start()

	def start(self):
		return self.engine
```

```python
Engine(4).start()
```

```python
Car("Tata",1500).start()
```

## ♨️ Class inside Class

```python
class Engine:
	def __init__(self, horsepower):
		self.horsepower = horsepower

	def start(self):
		return f"{self.horsepower} HP Engine Started"

class Car:
	class Features:
		def __init__(self, colour):
			self.colour = colour

	def __init__(self, brand, horsepower):
		self.brand = brand
		self.horsepower = horsepower
		self.engine = Engine(self.horsepower).start()
		self.colour = self.colour()

	def start(self):
		return self.engine

	def colour(self, colour = "red"):
		return self.Features(colour).colour
```

```python
Car("Tata", 1300).colour
```

### 👨‍🏫 Concept-2 : IS-A vs HAS-A (kab kya use kare?)

| Inheritance (IS-A) |Composition (HAS-A) |
|---|---|
| `Dog` IS-A `Animal` | `Car` HAS-A `Engine` |
| Type ka Rishta | Parts ka Rishta |
| `class Dog(Animal)` | `self.engine = Engine()` |
*"Rule:- Agar 'X ek tarah ka Y hai' -> inheritance. Agar 'X ke pass Y hai'-> Composition. Agent ke pass tools HOTE hain (has-a), toh Composition perfect hai."*
