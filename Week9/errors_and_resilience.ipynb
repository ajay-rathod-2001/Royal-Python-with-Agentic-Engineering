### 👨‍🏫 Concept-3: custom exception class

```python
class InsufficientFundError(Exception):
	"Raised when withdraw exceeds the balance"
	pass
```

```python
def withdraw_good(balance, amount):
	if amount > balance:
		raise InsufficientFundError(f"Can not withdraw: {amount}, current balance is: {balance}.")

	return balance -amount
```

```python
withdraw_good(500,1000)
```

```python
try:
	withdraw_good(500,1000)
except InsufficientFundError as e:
	print(f"Error: {e}") 
```

### 👨‍🏫 Concept-4: exception hierarchy (aone errors ka family)

```python
class ToolError(Exception):
	"Base error for the all tools problem"
	pass
```

```python
class ToolNotFoundError(Exception):
	"XYZ tool not found"
	pass
```

```python
class ToolTimeOutError(Exception):
	"Taking long time to get tools"
	pass
```

```python
try:
	tool_time = "Get current time tool"
	time_taken = "1100 seconds"
	raise ToolTimeOutError(f"Taking a long time[{time_taken}] to call {tool_time}")
except ToolTimeOutError as e:
	print(f"Error: {e}")
```

# 🌈🔀 Retry Pattern :

### 👨‍🏫 Concept-1: retry ka basic idea

```python
import time
```

```python
def flaky_task(attempt):
	if attempt < 3:
		raise ConnectionError("Networh Error")
	return "Success!"

max_attempt = 3

for attempt in range(1, max_attempt+1):
	try:
		result = flaky_task(attempt)
		print(f"Attempt {attempt}: {result}")
		break
	except ConnectionError as e:
		print(f"Attempt {attempt} failed: {e}")

		if attempt < max_attempt:
			time.sleep(1)
		else:
			print("All attempt failed, Giving up")
```

```python
flaky_task()
```

### 👨‍🏫 Concept-2: backoff (har bar zyada ruko)

```python
for attempt in range(1, 4):
	wait = 2 ** (attempt -1)
	time.sleep(wait)

	print(f"Attempt {attempt}, would wait {wait}s on failure")
```
